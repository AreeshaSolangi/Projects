```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Downloads/dirty_customer_dataset_100k.csv")
```

```python
df.shape
```

```python
df.isnull().sum()
```

# Remove Duplicates


```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.shape
```

## Split first and last name

```python
df = df.copy()
df[['First_name', 'Last_name']] = df['Full_Name'].str.split(n=1, expand=True)
```

```python
df.drop(columns='Full_Name',inplace=True)
```

```python
df=df[['CustomerID','First_name',  'Last_name','Age', 'Gender', 'Country', 'City', 'Email', 'Phone',
       'Signup_Date', 'Last_Purchase_Date', 'Total_Purchase_Amount',
       'Num_Orders', 'Product_Category', 'Rating', 'Loyalty_Tier',
       'Annual_Income', 'Discount_Used', 'Churn_Status' ]]
```

```python
df["Age"]=df["Age"].astype("Int64")
```

```python
df.dtypes
```

## Fix Age 

```python
df[(df["Age"]<18) | (df["Age"]>80)].shape[0]
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df.loc[(df["Age"] < 18) | (df["Age"] > 80), "Age"] = np.nan
```

```python
df["Age"].isnull().sum()
```

## Clean Gender

```python
df.loc[(df["Gender"].str.lower()=='female') | (df["Gender"].str.lower()=='f') ,"Gender"]='F'
df.loc[(df["Gender"].str.lower()=='male') | (df["Gender"].str.lower()=='m') ,"Gender"]='M'
```

```python
df[~df["Gender"].isin(['F','M'])]
```

## Clean Country

```python
df["Country"] = df["Country"].str.strip()
```

```python
df.loc[(df["Country"].str.lower()=="uk") |(df["Country"].str.lower()=="united kingdom"),"Country"]='UK'
df.loc[(df["Country"].str.lower()=="pk") |(df["Country"].str.lower()=="pakistan"),"Country"]='PK'
df.loc[(df["Country"].str.lower()=="usa") | (df["Country"].str.lower()=="united states") ,"Country"]='USA'
```

```python
df[~df["Country"].isin(['USA','UK','PK','GB'])]
```

## Fix Invaid Emails

```python
df["Email"]=df["Email"].str.strip()
```

```python
df.loc[df["Email"].str.contains("@@", na=False),"Email"]=np.nan
df.loc[~df["Email"].str.contains("@", na=False),"Email"]=np.nan
```

```python
df["Email"].isnull().sum()
```

## Fix Invalid Phone Numbers

```python
df["Phone"]=df["Phone"].str.strip()
```

```python
df["Phone"] = df["Phone"].str.replace(r"[-./\\]", "", regex=True)
df["Phone"] = df["Phone"].str.split("x").str[0]
```

```python
df["Phone"]=df["Phone"].str.strip()
```

```python
df.loc[df["Phone"].str.contains(r"[a-zA-Z]", na=False), "Phone"] = np.nan
```

# Fix invalid Future Signup Dates

```python
df["Signup_Date"]=pd.to_datetime(df["Signup_Date"])
```

```python
df["Last_Purchase_Date"]=pd.to_datetime(df["Last_Purchase_Date"])
```

## Clean Amount Values

```python
df["Total_Purchase_Amount"]=df["Total_Purchase_Amount"].str.replace(r"[Rs.£$USD]","", regex=True)
df["Total_Purchase_Amount"]=df["Total_Purchase_Amount"].astype("float")
df.loc[df["Total_Purchase_Amount"]<0,"Total_Purchase_Amount"]=np.nan
```

```python
df["Num_Orders"]=df["Num_Orders"].str.replace(r"[^\d]","", regex=True)
df["Num_Orders"] = df["Num_Orders"].replace("", np.nan)
df["Num_Orders"]=df["Num_Orders"].astype("float")
```

```python
df.loc[df["Num_Orders"]<0,"Num_Orders"]=np.nan
```

## Fix/Encode All Categorical Values

```python
df["Product_Category"].unique()
```

```python
df["Product_Category"] = df["Product_Category"].str.strip().str.title()
```

```python
df["Product_Category"].unique()
```

```python
df["Rating"]=df["Rating"].astype(float)
```

```python
df["Loyalty_Tier"] = df["Loyalty_Tier"].str.strip().str.title()
```

```python
df["Loyalty_Tier"].unique()
```

```python
df["Annual_Income"]=df["Annual_Income"].astype("float")
```

```python
df.loc[df["Annual_Income"]<0,"Annual_Income"]=np.nan
```

```python
df["Discount_Used"]=df["Discount_Used"].str.strip().str.title()
```

```python
df["Discount_Used"]=df["Discount_Used"].str.replace("1","Yes")
df["Discount_Used"]=df["Discount_Used"].str.replace("True","Yes")
df["Discount_Used"]=df["Discount_Used"].str.replace("0","No")
df["Discount_Used"]=df["Discount_Used"].str.replace("False","No")
```

```python
df["Churn_Status"]=df["Churn_Status"].str.strip().str.title()
```

```python
df["Churn_Status"]=df["Churn_Status"].str.replace("Churned","1")
df["Churn_Status"]=df["Churn_Status"].str.replace("Active","0")
```

```python
today=pd.Timestamp.today()
```

```python
df.loc[df["Signup_Date"] > today, "Signup_Date"] = pd.NaT
```

```python
df.loc[df["Last_Purchase_Date"] > today, "Last_Purchase_Date"] = pd.NaT
```

## Project Summary

This notebook documents a complete data cleaning workflow performed on a synthetic 
customer dataset of 101,500 rows and 18 columns, replicating real-world data quality 
issues commonly encountered in business environments.

### What Was Done

| Step | Action | Rows/Values Affected |
|---|---|---|
| Duplicate removal | Dropped exact duplicate records | 1,500 rows removed |
| Invalid ages | Replaced values outside 18–80 with NaN, imputed with median | 500+ values |
| Inconsistent categories | Standardized Gender, Country, Loyalty, Category, Discount, Churn | All string columns |
| Malformed emails | Replaced invalid formats with NaN | 600+ values |
| Currency symbols | Stripped $, £, Rs., USD using regex, converted to float | 350 values |
| Negative amounts | Replaced with NaN | 300 values |
| Wrong data types | Cleaned non-numeric strings in Num_Orders, converted to float | 400 values |
| Future dates | Replaced 2027 signup dates with NaT | 200 values |
| Feature engineering | Split Full_Name into First_name and Last_name | All rows |
| Label encoding | Encoded Churn_Status: Active = 0, Churned = 1 | All rows |

### Intentional Remaining Nulls

Rating nulls (17,634) were left as NaN deliberately. A missing rating means the 
customer did not rate, which is meaningful information. Imputing a rating value 
would misrepresent customer behaviour and introduce bias into any downstream analysis.

Email and phone nulls were also retained as NaN. Contact information cannot be 
estimated or imputed.

### Final Dataset

- **Rows:** 100,056  
- **Columns:** 19  
- **Status:** Clean and analysis-ready
