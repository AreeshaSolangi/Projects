# page 1 Scraping

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
options = Options()
options.add_argument("--headless=new")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--window-size=1920,1080")
options.add_argument("accept-language=en-US,en;q=0.9")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                     "AppleWebKit/537.36 (KHTML, like Gecko) "
                     "Chrome/120.0.0.0 Safari/537.36")

driver = webdriver.Chrome(options=options)
driver.get("https://www.amazon.com/s?k=maybelline+mascara&currency=USD")
links = driver.find_elements(By.XPATH, '//div[@class="a-section a-spacing-none a-spacing-top-small s-title-instructions-style"]/a')
product_links = [link.get_attribute("href") + "?currency=USD" for link in links]
print(f"🔗 Total product links found: {len(product_links)}")
titles = []
prices = []
ratings = []
reviews = []
availabilities = []
sellers = []  
for link in product_links:
    try:
        driver.get(link)
        wait = WebDriverWait(driver, 10)
        title = wait.until(EC.presence_of_element_located((By.XPATH, '//span[@id="productTitle"]'))).text
        titles.append(title)
        try:
            price = wait.until(EC.presence_of_element_located(
                (By.XPATH, '//span[contains(@class, "a-offscreen")]')
            )).get_attribute("textContent")
        except:
            price = "N/A"
        prices.append(price)
        try:
            rating = driver.find_element(By.XPATH, '//span[@data-hook="rating-out-of-text"]').text
        except:
            rating = "N/A"
        ratings.append(rating)
        try:
            review = driver.find_element(By.XPATH, '//span[@id="acrCustomerReviewText"]').text
        except:
            review = "N/A"
        reviews.append(review)
        try:
            availability = driver.find_element(By.XPATH, '//div[@id="availability"]//span').text
        except:
            availability = "N/A"
        availabilities.append(availability)
        try:
            seller_text = driver.find_element(By.XPATH, '//a[contains(@href, "offer-listing")]').text
            sellers.append(seller_text)
        except:
            sellers.append("N/A")

    except Exception as e:
        print(f"❌ Error fetching {link[:60]}... → {e}")
        titles.append("N/A")
        prices.append("N/A")
        ratings.append("N/A")
        reviews.append("N/A")
        availabilities.append("N/A")
        sellers.append("N/A")

# --- Step 5: Create DataFrame ---
df = pd.DataFrame({
    "Title": titles,
    "Price (USD)": prices,
    "Rating": ratings,
    "Reviews": reviews,
    "Availability": availabilities,
    "Number of Sellers": sellers,
    "Product Link": product_links
})

driver.quit()
```

# scraping multiple pages

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time

# --- Browser setup ---
options = Options()
options.add_argument("--headless=new")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--window-size=1920,1080")
options.add_argument("accept-language=en-US,en;q=0.9")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                     "AppleWebKit/537.36 (KHTML, like Gecko) "
                     "Chrome/120.0.0.0 Safari/537.36")

driver = webdriver.Chrome(options=options)

# --- Step 1: Open Amazon homepage and force USD ---
driver.get("https://www.amazon.com/")
driver.add_cookie({"name": "i18n-prefs", "value": "USD"})
driver.add_cookie({"name": "lc-main", "value": "en_US"})
time.sleep(1)

# --- Step 2: Open search page ---
search_url = "https://www.amazon.com/s?k=maybelline+mascara&currency=USD"
driver.get(search_url)

# --- Step 3: Wait and get all pagination links (raw URLs) ---
try:
    WebDriverWait(driver, 10).until(
        EC.presence_of_all_elements_located((By.XPATH, '//a[contains(@aria-label, "Go to page")]'))
    )
    pagination_links = [
        link.get_attribute("href") for link in driver.find_elements(By.XPATH, '//a[contains(@aria-label, "Go to page")]')
    ]
    pagination_links = list(dict.fromkeys(pagination_links))  # remove duplicates
    print(f"📄 Total pagination pages found: {len(pagination_links)}")
except Exception as e:
    print(f"⚠️ Could not find pagination: {e}")
    pagination_links = [search_url]

# --- Step 4: Prepare lists ---
titles, prices, ratings, reviews, availabilities, sellers, product_links_all = [], [], [], [], [], [], []

# --- Step 5: Loop through each pagination link ---
for page_link in pagination_links:
    print(f"\n🌍 Scraping Page: {page_link}")
    driver.get(page_link)

    # Wait for product links
    try:
        WebDriverWait(driver, 10).until(
            EC.presence_of_all_elements_located((By.XPATH, '//div[@class="a-section a-spacing-none a-spacing-top-small s-title-instructions-style"]/a'))
        )
    except:
        print("⚠️ No product links found on this page, skipping.")
        continue

    # Collect product links (raw)
    product_links = [
        link.get_attribute("href")
        for link in driver.find_elements(By.XPATH, '//div[@class="a-section a-spacing-none a-spacing-top-small s-title-instructions-style"]/a')
    ]
    print(f"🔗 Products found: {len(product_links)}")

    # --- Inner loop: visit each product ---
    for link in product_links:
        try:
            driver.get(link)
            wait = WebDriverWait(driver, 10)

            # --- Title ---
            title = wait.until(EC.presence_of_element_located((By.XPATH, '//span[@id="productTitle"]'))).text
            titles.append(title)

            # --- Price ---
            try:
                price = wait.until(EC.presence_of_element_located(
                    (By.XPATH, '//span[contains(@class, "a-offscreen")]')
                )).get_attribute("textContent")
                if "PKR" in price:  # Fix if any PKR appears
                    price = price.replace("PKR", "$")
            except:
                price = "N/A"
            prices.append(price)

            # --- Rating ---
            try:
                rating = driver.find_element(By.XPATH, '//span[@data-hook="rating-out-of-text"]').text
            except:
                rating = "N/A"
            ratings.append(rating)

            # --- Reviews ---
            try:
                review = driver.find_element(By.XPATH, '//span[@id="acrCustomerReviewText"]').text
            except:
                review = "N/A"
            reviews.append(review)

            # --- Availability ---
            try:
                availability = driver.find_element(By.XPATH, '//div[@id="availability"]//span').text
            except:
                availability = "N/A"
            availabilities.append(availability)

            # --- Sellers ---
            try:
                seller_text = driver.find_element(By.XPATH, '//a[contains(@href, "offer-listing")]').text
                seller_text = seller_text.replace("\n", "").strip()
                seller_text = seller_text.split("from")[0].strip() if "from" in seller_text else seller_text
            except:
                seller_text = "N/A"
            sellers.append(seller_text)

            product_links_all.append(link)

        except Exception as e:
            print(f"❌ Error fetching {link[:60]} → {e}")
            titles.append("N/A")
            prices.append("N/A")
            ratings.append("N/A")
            reviews.append("N/A")
            availabilities.append("N/A")
            sellers.append("N/A")
            product_links_all.append(link)

# --- Step 6: Create DataFrame ---
df = pd.DataFrame({
    "Title": titles,
    "Price (USD)": prices,
    "Rating": ratings,
    "Reviews": reviews,
    "Availability": availabilities,
    "Number of Sellers": sellers,
    "Product Link": product_links_all
})

# --- Step 7: Display summary ---
pd.set_option('display.max_rows', None)
print("\n📊 Scraping Completed!")
print(f"Total products scraped: {len(df)}")
print(df[["Title", "Price (USD)"]].head())

driver.quit()

```
